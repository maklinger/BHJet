## Parameters in this setup

The parameters listed below are a subset of those used in BHJet and the most relevant for getting to know how to use it. These are either source-specific or parameters that we often change while using the model & fitting SEDs. Parameters beginning with `lg_10_` are shown as base-10 logarithms.

You can use the sliders to change the values, or you can also click on the first row of numbers and directly put in values. (Sometimes a little finnicky!)

### Source and viewing geometry

- `lg_10_mass_bh` — black-hole mass in solar masses. It sets the gravitational length scale, $r_g = GM/c^2$.
- `theta_obs` — viewing angle between our line of sight and the jet axis. 
- `lg_10_distance` — source distance in kpc.

### Jet structure and dynamics

- `lg_10_r_initial` — jet radius at its base, in $r_g$.
- `lg_10_jet_power_eddington` — jet power in units of Eddington luminosity (theoretical maximum); this is a main control on the overall energy available to the jet.
- `lg_10_z_dissipation` — distance where particles begin to be accelerated into a non-thermal distribution, in $r_g$.
- `lg_10_z_end_of_acceleration` — distance where jet (bulk) acceleration ends, in $r_g$.
- `lg_10_sigma_final` — final magnetisation, describing the balance between magnetic and particle energy at the distance of z_end_of_acceleration.
- `lg_10_gamma_final` — maximum bulk Lorentz factor of the flow; the jet reaches this value at z_end_of_acceleration. 
- `lg_10_z_max_calculation` — maximum length of the jet.

### Radiating electrons

- `lg_10_electron_temperature_jet_base` — thermal electron temperature at the jet base, in keV.
- `lg_10_fraction_nonthermal_electrons` — fraction of electrons accelerated into the non-thermal population. 
- `lg_10_factor_break_electrons` — factor controlling the electron break energy.
- `lg_10_factor_max_energy_electrons` — parameter controlling the maximum electron energy. The value corresponds to a factor in the equation that balances the cooling and acceleration timescales of the parameters
- `index_injected_electrons` — power-law index of the accelerated electron distribution; it changes the non-thermal spectral slope.
